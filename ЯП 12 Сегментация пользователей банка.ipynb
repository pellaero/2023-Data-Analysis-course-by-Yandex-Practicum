### 2.3. Проверка на пропуски

```python
# выявление пропусков данных в разрезе каждого признака в абсолютных и относительных значениях
bank_missing = pd.DataFrame(columns=['название признака', 'количество пропусков', '% пропущенных значений'])
j = 0
for i, col in enumerate(bank.columns):
    percent_missing = np.mean(bank[col].isnull())
    value_missing = np.sum(bank[col].isnull())
    if percent_missing != 0:
        bank_missing.loc[j] = [col, value_missing, (percent_missing*100).round(2)]
        j +=1

print('Таблица 2. Информация о количестве пропусков', '\n\n', bank_missing.sort_values(by='количество пропусков', ascending=True))
```

### 2.4. Обработка пропусков

Ранее нами было установлено наличие пропусков в данных по признакам `age` и `balance`. Пропуски по признаку `age` невозможно восстановить по целому ряду причин: персональные данные клиента не могут быть восстановлены без потери качества информации, данные не могут быть восстановлены путём подставления среднего либо медианного значения. На основании количественных характеристик пропуска по указанному признаку полагаем возможным удалить эти строки.

```python
# удаляем строки, где пропущены значения в признаке `age`
bank = bank.dropna(subset=['age'])
```

Выбор коэффициента корреляции для расчёта матрицы значений был сделан в пользу $𝜙k$, поскольку он основан на информационной теории и представляет собой расширение коэффициента корреляции $𝜙$ (Phi) для случая с несколькими категориями.

Коэффициент корреляции `Phik` принимает значения от 0 до 1. Значение 1 указывает на полную положительную связь между переменными, а значение 0 указывает на отсутствие связи.

Также учитвыется информация о порядке категорий, может быть обнаружена нелинейная связь между переменными. Это делает коэффициент $𝜙k$ более гибким в сравнении с другими коэффициентами корреляции, такими как коэффициент `V Крамера`. В нашем наборе данных много категориальных переменных, поэтому, полагаем оптимальным использование коэффициента $𝜙k$.

Справочный материал, на основании которого мы пришли к выбору коэффициента корреляции:
* https://toto-school.ru/raznoe-2/koefficzient-kramera-koefficzienty-korrelyaczii.html
* https://towardsdatascience.com/phik-k-get-familiar-with-the-latest-correlation-coefficient-9ba0032b37e7

```python
# расчёт корреляции с помощью коэффициента Phik
corr_Phik = balance_corr.phik_matrix()
```

```python
# визуализация матрицы корреляций по критерию `Phic`
fig, ax = plt.subplots(figsize=(12,12))
ax = sns.heatmap(data = corr_Phik, annot=True, square=True, center=0,
            cmap='coolwarm', fmt='.2f', linewidths=0.5, linecolor='black', mask=np.triu(corr_Phik))
plt.title('Матрица корреляций с коэффициентом Phik')
plt.show()
```

На основании матрицы полученных значений статистической меры силы и направления связи между двумя категориальными переменными, мы можем определить перечень признаков, наиболее сильно влияющих на признак `balance`.

```python
# мера взаимосвязи между признаком `balance` и другими признаками
corr_Phik['balance'].sort_values(ascending=False)[1:]
```

Таким образом, расчёт матрицы корреляции с коэффициентом `Phik` показывает наличие умеренной связи признака `balance` с признаком `estimated_salary` и слабой связи с признаком `score`. 

На основании установленных взаимосвязей признака `balance` с признаками `estimated_salary` и  `score` мы можем утверждать, что имеющиеся пропуски по признаку `balance` были сформированы посредством механизма `MAR (Missing At Random)`, где пропуск данных объясняется некоторой закономерностью, которая определеяется на основе другой имеющейся в наборе данных информации (в нашем случае это кредитный скор и оценочная заработная плата клиента). В этом случае, обработка и замена пропусков в признаке `balance` может быть выполнена на основе двух вышеназванных признаков без существенного искажения результатов.

Ранее нами была установлена умеренная взаимосвязь между признаком `estimated_salary` и слабая взаимосвязь между признаком `score` и целевым признаком `balance`. Одновременно с этим, в признаке `balance` имеются пропуски. Для заполнения пропущенных значений на основе указанных признаков мы можем применить разные методы. Нами был реализован метод мультипликативной импьютации (multiplicative imputation) или ковариационной импьютации (covariate imputation), учитывающий взаимосвязь указанных признаков. У такого метода есть свои преимущества и недостатки.

Достоинства метода мультипликативной импьютации:
* Учитывает зависимости между переменными: метод мультипликативной импьютации учитывает корреляцию между переменными при заполнении пропущенных значений. Это позволяет сохранить связи между переменными и более реалистично восстановить пропущенные данные.
* Подходит для данных с нелинейными зависимостями: мультипликативная импьютация хорошо работает в случаях, когда зависимость между переменными является нелинейной. Она учитывает не только линейные связи, но и возможные нелинейности в данных.
* Сохраняет дисперсию: применение мультипликативной импьютации позволяет сохранить дисперсию восстановленных данных. Это полезно, если дисперсия является важным аспектом анализа данных.

Недостатки метода мультипликативной импьютации:
* Может привести к искажению данных: в ряде случаев, особенно когда данные содержат выбросы или аномальные значения, мультипликативная импьютация может привести к искажению данных. Заполнение пропущенных значений на основе других переменных может изменить распределение данных и сделать его менее точным.
* Требует предварительного анализа и моделирования: применение метода мультипликативной импьютации требует тщательного анализа и моделирования зависимостей между переменными. Если модель неправильно учитывает зависимости, то импьютация может дать неточные результаты.

```python
# создаём копию оригинального датафрейма
df = bank.copy()

# делим исходные данные две части: с пропущенными значениями и без пропущенных значений в столбце 'balance'
df_missing = df[df['balance'].isnull()]
df_not_missing = df[~df['balance'].isnull()]

# создаём матрицу X и вектор y для обучения модели
X = df_not_missing[['estimated_salary', 'score']]
y = df_not_missing['balance']

# обучаем модель линейной регрессии на имеющихся значениях
model = LinearRegression()
model.fit(X, y)

# используем обученную модель для предсказания пропущенных значений
X_missing = df_missing[['estimated_salary', 'score']]
predicted_balance = model.predict(X_missing)

# заполняем пропуски значений `balance` предсказанными значениями
df.loc[df['balance'].isnull(), 'balance'] = predicted_balance
```

```python
# создаём гистограмму для визуализации полученных результатов заполнения пропусков
fig, ax = plt.subplots(figsize=(12, 6))

# на графике выводим совмещённое распределение данных с пороговым значением в 4 000 000 руб.
# оригинальный массив данных, распределение значений признака `balance`
sns.histplot(bank.query('balance < 4000000')['balance'], stat='density', 
             common_norm=False, bins=100, ax=ax, color='green', label='оригинальный датафрейм')

# заполненный массив данных, распределение значений признака `balance`
sns.histplot(df.query('balance < 4000000')['balance'], stat='density', 
             common_norm=False, bins=100, ax=ax, color='#FF8B00', alpha=0.6, label='заполненный датафрейм')
ax.set_title('Распределение значений признака `balance` \n на оригинальном и заполненном массивах данных', fontsize=14)
plt.ylabel('плотность значений', fontsize=12)
plt.xlabel('баланс счёта', fontsize=12)
plt.legend()
plt.show()
```

### 3.1. Описательная статистика: вычисление основных статистических показателей (среднее, медиана, 95-й и 99-й перцентиль) для числовых данных клиентов в разрезе количества продуктов:
- формирование групп клиентов в соответствии с количеством используемых продуктов;
- расчёт описательных статистик вышеназванных групп.

В нашем распоряжении имеется массив данных, в котором обозначены 5 банковских продуктов, наименование и содержание которых доподлинно нам неизвестно. Предлагаем методом срезов `query` исследовать количество клиентов, которые предпочитают пользоваться тем или иным количеством продуктов.

Коэффициент вариации позволит нам оценить относительную изменчивость данных в процентах от среднего значения. Для этого рассчитаем относительную вариацию (коэффициент вариации) для каждого набора данных. Разделим стандартное отклонение (показатель разброса данных) на разность между средним значением и минимальным значением данных.

```python
# функция для вывода описательных статистик в разрезе количества банковских продуктов у клиента 
def calculate_group_stats(df, group_number):
    # список признаков для расчёта
    columns_num = ['user_id', 'age', 'balance', 'estimated_salary', 'score', 'equity', 'products']

    gr = df.query(f'products == {group_number}')
    gr_df = gr.describe()
    
    # считаем коэффициент вариации для конкретной группы
    coef_variation = gr[columns_num].std(numeric_only=True) / (gr[columns_num].mean(numeric_only=True) - gr[columns_num].min(numeric_only=True))

    for column in columns_num:
        gr_df.loc['median', column] = gr[column].median()
        gr_df.loc['95%', column] = gr[column].quantile(0.95)
        gr_df.loc['99%', column] = gr[column].quantile(0.99)
        gr_df.loc['coef_variation', column] = coef_variation[column]

    # изменяем порядок индекса
    new_index = ['count', 'coef_variation', 'mean', 'std', 'median', 'min', '25%', '50%', '75%', '95%', '99%', 'max']
    gr_df = gr_df.reindex(new_index)
    
    # выводим описательные статистики для указанного количества продуктов
    display(f'Описательные статистики по клиентам, пользующимся {group_number} банковскими продуктами:')
    display(gr_df)
    display(f'Численные показатели по категориальным признакам в этой группе:')
    display(df.query(f'products == {group_number}')['gender'].\
            replace({True: 'мужчин', False: 'женщин'}).value_counts().to_frame())
    display(df.query(f'products == {group_number}')['city'].value_counts().to_frame())
    display(df.query(f'products == {group_number}')['credit_card'].\
            replace({True: 'есть кредитная карта', False: 'нет кредитной карты'}).value_counts().to_frame())
    display(df.query(f'products == {group_number}')['last_activity'].\
            replace({True: 'активных клиентов', False: 'неактивных клиентов'}).value_counts().to_frame())
    display(df.query(f'products == {group_number}')['churn'].\
            replace({True: 'ушедших клиентов', False: 'оставшихся клиентов'}).value_counts().to_frame())
```

### 3.2.15. Какова связь между скоринговым баллом клиента и его балансом? regplot

Для исследования в этом блоке предлагаем сделать срез данных по 99-му перцентилю значений признака `balance`:

```python
# изучим значения признака `balance` для последовательного установления срезов данных для графиков
df.balance.describe()
```

```python
# разница между самым большим значением признака `balance` и 99-м перцентилем весьма внушительная
df.balance.max() - df['balance'].quantile(0.99).round(2)
```

```python
# срез по 99-му перцентилю
bal_common = df.query('balance < 5682756.54')
```

```python
# разделим значения признака balance по квантилям
bal_common['balance_quantile'] = pd.qcut(bal_common['balance'], q=4)
bal_common['balance_quantile'] = bal_common['balance_quantile'].astype('str')
quantiles = bal_common['balance_quantile'].unique()

# выводим графики `regplot` для каждого квантиля
for quantile in quantiles:
    plt.figure(figsize=(12, 4))
    sns.regplot(x='score', y='balance', data=bal_common.query(f"balance_quantile == '{quantile}'"),
                color='brown', scatter_kws=dict(color='orange'))
    plt.title(f'Взаимосвязь баланса счёта и кредитного скора клиента в квантиле {quantile}', fontsize=14)
    plt.ylabel('баланс', fontsize=12)
    plt.xlabel('кредитный скор', fontsize=12)
    plt.show()
```

**Вывод:** по расчётам матрицы корреляции была установлена слабая положительная мера связи признаков `balance` и `score`. Показатель меры связи равен 0.17. Это достаточно низкое значение, чтобы говорить об определяющей роли одного признака в формировании значений другого.

Для визуального подтверждения линейной положительной зависимости признаков мы с помощью метода `pd.qcut()` разделили признак `balance` на квантили и построили графики регрессионной зависимости на каждом участке. На третьем и четвёртом участке квантилей линия доверительного интервала вокруг линии регрессии самая широкая. Ширина этой полосы указывает на степень неопределённости и разброса в имеющихся у нас данных. Следовательно, имеется большая неопределенность в оценках признаков `balance` и `score`. 

На этих участках квартилей результаты модели линейной регрессии менее надежны. И наоборот, узкая полоса доверительного интервала на первом и втором участках указывает нам на большую уверенность в оценках признаков и более точные результаты.

### 3.2.20. Взаимосвязь уникального номера клиента и города, в котором клиент проживает (косвенная информация о датах создания головного офиса и его филиалов, потенциально – масштабирование стандартов обслуживания клиентов и, как следствие, отток в случае неудовлетворённости качеством) гистограмма

```python
# функция для вывода гистограммы 
def plot_density(data, title):
    unique_cities = data['city'].unique()
    
    for city in unique_cities:
        churn_true_count = len(data[(data['city'] == city) & (data['churn'] == True)])
        churn_false_count = len(data[(data['city'] == city) & (data['churn'] == False)])
        
        fig, ax = plt.subplots(figsize=(12, 4))
        plt.title(f'{title}, город {city}', fontsize=14)
        sns.histplot(data=data[data['city'] == city], x='user_id', hue='churn', stat='density', bins=100, common_norm=False,
                     palette=['#cc6c5c', '#69a1ea'])
        plt.axvline(x=115000, color='red', linestyle='--', linewidth=3)
        plt.xlabel('сквозные уникальные идентификационные номера клиентов', fontsize=12)
        plt.ylabel('плотность значений', fontsize=12)
        plt.text(0.65, 0.9, f'клиент ушёл: {churn_true_count} человек', transform=plt.gca().transAxes)
        plt.text(0.65, 0.85, f'клиент остался: {churn_false_count} человек', transform=plt.gca().transAxes)
        plt.legend(loc='upper left', labels=['постоянные и новые клиенты', 'клиент остался', 'клиент ушёл'])
        plt.show()

plot_density(df, 'Плотность распределения значений признака: все клиенты банка')
plot_density(df[df.user_id <= 115000], 'Плотность распределения значений признака: постоянные клиенты')
plot_density(df[df.user_id > 115000], 'Плотность распределения значений признака: новые клиенты')
```

### 3.3.7. Какая доля неактивных клиентов уходит/остаётся в разрезе количества используемых банковских продуктов?

```python
# создаём пустой словарь для хранения результатов
activity_churn_ratio = {}

# циклом проходим по количеству продуктов от 1 до 5
for num_products in range(1, 6):
    current_data = df[df['products'] == num_products]
    total_clients = len(current_data)
    churn_true_clients = current_data.query('last_activity==False')[current_data['churn'] == True]['user_id'].count()
    churn_false_clients = current_data.query('last_activity==False')[current_data['churn'] == False]['user_id'].count()
    churn_true_ratio = churn_true_clients / total_clients
    churn_false_ratio = churn_false_clients / total_clients
    activity_churn_ratio[num_products] = {'churn_true_ratio': churn_true_ratio, 'churn_false_ratio': churn_false_ratio}

# из словаря `balance_ratio` создаём датафрейм
activity_churn_ratio_df = pd.DataFrame(activity_churn_ratio).transpose()
```

###  4.1. Проверка статистической гипотезы о различии распределений возраста между теми клиентами с одним и двумя продуктами.

**Сформулируем гипотезу:**

$H_{0}: \text{распределение возраста клиентов с 1 продуктом} = \text{распределение возраста клиентов с 2 продуктами}$
$H_{1}: \text{распределение возраста клиентов с 1 продуктом} \neq \text{распределение возраста клиентов с 1 продуктом}$

Установим пороговое значение:
$ \alpha = 0.05$

**Выбор критерия:** Чтобы уменьшить неопределённость относительно интересующего нас вопроса, мы можем провести двусторонний тест, поскольку у нас есть необходимость проверить альтернативное существование неравенства, будь оно положительным или отрицательным. Для проверки гипотезы о равенстве распределений возраста двух независимых групп мы используем **критерий Манна-Уитни (непараметрический тест),** поскольку:

* Распределения возраста в обеих группах не являются нормальными (проверено в Шаге 3);
* Критерий Манна-Уитни сравнивает распределения (ранги), а не средние значения;
* Тест устойчив к выбросам и не требует предположения о нормальности.

```python
# создаём выборки клиентов с одним и двумя продуктами
age_one_product = df.query('products == 1')['age']
age_two_products = df.query('products == 2')['age']
```

```python
plt.figure(figsize=(10, 5))
sns.kdeplot(age_one_product, fill=True, label='1 продукт', color='blue', alpha=0.5)
sns.kdeplot(age_two_products, fill=True, label='2 продукта', color='orange', alpha=0.5)
plt.title('Сравнение распределений возраста клиентов', fontsize=14)
plt.xlabel('Возраст клиента')
plt.ylabel('Плотность вероятности')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
# ящик с усами для выборок: средний возраст клиентских групп
plt.figure(figsize=(12, 4))

plt.subplot(2, 1, 1)
plt.title('Размах значений возраста клиентов с 1 продуктом', fontsize=14)
sns.boxplot(x=age_one_product.reset_index(drop=True), showmeans=True, color='#FFB840')
plt.ylabel('размах значений', fontsize=12)
plt.xlabel('возраст', fontsize=12)

plt.subplot(2, 1, 2)
plt.title('Размах значений возраста клиентов с 2 продуктами', fontsize=14)
sns.boxplot(x=age_two_products.reset_index(drop=True), showmeans=True, color='#FF9073')
plt.ylabel('размах значений', fontsize=12)
plt.xlabel('возраст', fontsize=12)

plt.tight_layout()
plt.show()
```

Проверим имеющиеся данные на соответствие вышеперечисленным параметрам.
<br>1. Независимость выборок.

В нашем исследовании выборки содержат разные наблюдения, относящиеся к разным клиентам, которые различаются с помощью групповой переменной, относящейся к шкале отношений, такой, как возраст клиента. Закономерное и однозначное соответствие между выборками невозможно, следовательно, эти выборки являются независимыми. 

2. Нормальность распределения: Для проверки нормальности распределения можно использовать функцию `normaltest` из модуля `scipy.stats`. Она основана на скосе и эксцессе распределения.

```python
# проверка нормальности распределения
statistic, p_value1 = stats.normaltest(age_one_product)
alpha = 0.05

if p_value1 < alpha:
    print('Распределение age_one_product не является нормальным.')
else:
    print('Распределение age_one_product является нормальным.')
    
statistic, p_value2 = stats.normaltest(age_two_products)
alpha = 0.05

if p_value2 < alpha:
    print('Распределение age_two_products не является нормальным.')
else:
    print('Распределение age_two_products является нормальным.')
```

3. Равенство дисперсий: Для проверки равенства дисперсий можно использовать функцию `var` из библиотеки `NumPy` для вычисления выборочной дисперсии. Параметр `ddof=1` указывает на использование поправки Бесселя для несмещенной оценки дисперсии.

```python
# сравнение дисперсии выборок
variance_one_product = np.var(age_one_product, ddof=1)
variance_two_products = np.var(age_two_products, ddof=1)

if variance_one_product > variance_two_products:
    print('Дисперсия выборки age_one_product больше, чем у age_two_products.')
elif variance_one_product < variance_two_products:
    print('Дисперсия выборки age_one_product меньше, чем у age_two_products.')
else:
    print('Дисперсии выборок age_one_product и age_two_products равны.')
```

Подытоживая результаты проверки параметров выборок, мы убеждаемся в необходимости применения статистического непараметрического `критерия Манна-Уитни`, который позволяет сравнить две независимые выборки без предположений о распределении данных. Данный критерий имеет широкую сферу применения в анализе данных, он используется для проверки статистических гипотез о различии между двумя независимыми выборками. Вот некоторые сферы, где он часто применяется:

* Медицинские исследования: для сравнения эффективности лечения или препаратов на основе результатов исследований на двух независимых группах пациентов.
* Социальные исследования: в социальных исследованиях критерий Манна-Уитни может быть использован для сравнения характеристик или поведения двух независимых групп людей.
* Бизнес-аналитика: может помочь в сравнении производительности или предпочтений клиентов в зависимости от разных факторов, таких как продукты или услуги, которые они используют.
* Экспериментальная психология: в экспериментах по психологии этот критерий может применяться для анализа результатов экспериментов, где исследуются две независимые группы испытуемых.
* Биологические исследования: может использоваться для сравнения биологических показателей или результатов экспериментов на двух независимых группах образцов.

В целом, критерий Манна-Уитни является мощным инструментом **для сравнения двух независимых выборок, когда данные не удовлетворяют требованиям параметрических тестов, таких как t-тест**. Он позволяет проводить статистические выводы о различиях между группами, не делая предположений о распределении данных. Этот тест основывается на рангах значений в выборках и тестирует гипотезу о равенстве средних значений или медиан.

Требования к данным для применения критерия Манна-Уитни:

* Две независимые выборки: клиенты, использующие один продукт, и клиенты, использующие два продукта.
* Значения признака (в нашем случае - возраст) должны быть измерены на уровне, который можно сравнить между собой (числовая шкала возраста клиента).
* Данные должны быть достаточно большими для получения статистически значимых результатов.

Проверим, соответствуют ли имеющиеся данные этим требованиям:

* У нас есть две выборки: клиенты, использующие один продукт, и клиенты, использующие два продукта.
* Возраст клиента хранится в признаке `df['age']`, что позволяет сравнивать значения между группами.
* Для проверки требуется достаточно большой объём данных в каждой выборке: 3246 значений и 5101 значений соответственно.

Поскольку, имеющиеся данные удовлетворяют этим требованиям, мы можем использовать критерий Манна-Уитни для проверки статистической гипотезы о различии распределений возраста между клиентами, использующими один и два продукта банка.

```python
# 1. две независимые выборки
if len(age_one_product) > 0 and len(age_two_products) > 0:
    # 2. значения возраста должны быть измерены на уровне, который можно сравнить между собой
    if age_one_product.dtype in ['float64', 'int64']:
        if age_two_products.dtype in ['float64', 'int64']:
                        
            # применение критерия Манна-Уитни
            stat, p_value = stats.mannwhitneyu(age_one_product, age_two_products, alternative='two-sided')
            alpha = 0.05
            
            print('Статистика критерия U:', stat)
            print('p-значение:', p_value)
            
            if p_value < alpha:
                print('Отвергаем нулевую гипотезу: распределения возраста клиентов с одним и двумя продуктами банка статистически различаются.')
            else:
                print('Не удалось отвергнуть нулевую гипотезу: распределения возраста клиентов с одним и двумя продуктами банка статистически неотличимы.')
        else:
            print('Требование не выполнено: тип данных возраста для выборки с двумя продуктами должен быть числовым.')
    else:
        print('Требование не выполнено: тип данных возраста для выборки с одним продуктом должен быть числовым.')
else:
    print('Требование не выполнено: выборки клиентов с одним и двумя продуктами должны содержать данные.')
```

**Вывод и бизнес-интерпретация:**
p-значение составило `0.061`, что превышает пороговое значение `α = 0.05`. Следовательно, у нас **нет статистических оснований отвергнуть нулевую гипотезу**. Распределения возраста клиентов с 1 и 2 продуктами статистически неотличимы.

**Ценность для бизнеса:** 
Возрастная структура клиентов с 1 и 2 продуктами однородна. Это означает, что количество продуктов не связано с возрастом клиента. Следовательно, **возраст не является значимым признаком (маркером)** для разделения этих двух групп при финальной сегментации. Мы не будем тратить ресурсы на создание разных возрастных стратегий для этих когорт.

### 4.2. Проверка статистической гипотезы о различии распределний количества баллов собственности между теми клиентами, которые продолжают пользоваться услугами банка, и теми, которые прекратили сотрудничество с банком.

**Сформулируем гипотезу:**

$H_{0}: \text{Распределение среднего балла собственности ушедших клиентов} = \text{Распределение среднего балла собственности оставшихся клиентов}$
$H_{1}: \text{Распределение среднего балла собственности ушедших клиентов} \neq \text{Распределение среднего балла собственности ушедших клиентов}$

Установим пороговое значение:
$ \alpha = 0.05$

**Выбор критерия:** Аналогично используем **критерий Манна-Уитни,** так как распределение балла собственности не является нормальным и содержит дискретные значения (от 0 до 9).

```python
# создаём выборки ушедших и оставшихся клиентов по признаку количества баллов собственности
eq_churn = df.query('churn == True')['equity']
eq_not_churn = df.query('churn == False')['equity']
```

```python
# парные диаграммы для выборок (дискретный признак equity)
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.title('Плотность распределения баллов \n собственности у ушедших клиентов', fontsize=14)
sns.histplot(eq_churn, stat='density', discrete=True, color='#679ED2', edgecolor='white')
plt.ylabel('плотность значений', fontsize=12)
plt.xlabel('балл собственности', fontsize=12)
plt.xticks(range(10)) # подписи 0, 1, 2 ... 9

plt.subplot(1, 2, 2)
plt.title('Плотность распределения баллов \n собственности у оставшихся клиентов', fontsize=14)
sns.histplot(eq_not_churn, stat='density', discrete=True, color='#FFD300', edgecolor='white')
plt.ylabel('плотность значений', fontsize=12)
plt.xlabel('балл собственности', fontsize=12)
plt.xticks(range(10))

plt.tight_layout()
plt.show()
```

Проверим имеющиеся данные на соответствие вышеперечисленным параметрам.
<br>1. Независимость выборок.

В этой части исследования мы рассматриваем наблюдения в отношении каждого клиента индивидуально. Эти наблюдения различаются с помощью групповой переменной, относящейся к шкале отношений, такой, как балл собственности. Закономерное и однозначное соответствие между выборками невозможно, следовательно, эти выборки являются независимыми. 

2. Нормальность распределения: Для проверки нормальности распределения можно использовать функцию `normaltest` из модуля `scipy.stats`. Она основана на скосе и эксцессе распределения.

```python
# проверка нормальности распределения
statistic, p_value1 = stats.normaltest(eq_churn)
alpha = 0.05

if p_value1 < alpha:
    print('Распределение eq_churn не является нормальным.')
else:
    print('Распределение eq_churn является нормальным.')
    
statistic, p_value2 = stats.normaltest(eq_not_churn)
alpha = 0.05

if p_value2 < alpha:
    print('Распределение eq_not_churn не является нормальным.')
else:
    print('Распределение eq_not_churn является нормальным.')
```

3. Равенство дисперсий: Для проверки равенства дисперсий можно использовать функцию `var` из библиотеки `NumPy` для вычисления выборочной дисперсии. Параметр `ddof=1` указывает на использование поправки Бесселя для несмещенной оценки дисперсии.

```python
# сравнение дисперсии выборок
variance_eq_churn = np.var(eq_churn, ddof=1)
variance_eq_not_churn = np.var(eq_not_churn, ddof=1)

if variance_eq_churn > variance_eq_not_churn:
    print('Дисперсия выборки eq_churn больше, чем у eq_not_churn.')
elif variance_eq_churn < variance_eq_not_churn:
    print('Дисперсия выборки eq_churn меньше, чем у eq_not_churn.')
else:
    print('Дисперсии выборок eq_churn и eq_not_churn равны.')
```

Проводим статистический непараметрический тест с использованием `критерия Манна-Уитни`, который позволяет сравнить две независимые выборки без предположений о распределении данных.

Требования к данным для применения критерия Манна-Уитни:

* Две независимые выборки: клиенты ушедшие и оставшиеся в разрезе количества баллов собственности.
* Значения признака (в нашем случае - балл собственности) должны быть измерены на уровне, который можно сравнить между собой (числовая шкала балла собственности).
* Данные должны быть достаточно большими для получения статистически значимых результатов.

Проверим, соответствуют ли имеющиеся данные этим требованиям:

* У нас есть две выборки: клиенты ушедшие и оставшиеся в разрезе количества баллов собственности.
* Балл собственности клиентов хранится в признаке `df['equity']`, что позволяет сравнивать значения между группами.
* Для проверки требуется достаточно большой объём данных в каждой выборке: 1817 значений и 8059 значений соответственно.

Поскольку, имеющиеся данные удовлетворяют этим требованиям, мы можем использовать критерий Манна-Уитни для проверки статистической гипотезы о различии возраста между клиентами, использующими один и два продукта банка.

```python
# 1. две независимые выборки
if len(eq_churn) > 0 and len(eq_not_churn) > 0:
    # 2. значения балла собственности должны быть измерены на уровне, который можно сравнить между собой
    if eq_churn.dtype in ['float64', 'int64']:
        if eq_not_churn.dtype in ['float64', 'int64']:
                        
            # применение критерия Манна-Уитни
            stat, p_value = stats.mannwhitneyu(eq_churn, eq_not_churn, alternative='two-sided')
            alpha = 0.05
            
            print('Статистика критерия U:', stat)
            print('p-значение:', p_value)
            
            if p_value < alpha:
                print('Отвергаем нулевую гипотезу: распределения балла собственности у ушедших и оставшихся клиентов статистически различаются.')
            else:
                print('Не удалось отвергнуть нулевую гипотезу: распределения балла собственности у ушедших и оставшихся клиентов статистически неотличимы.')
        else:
            print('Требование не выполнено: тип данных балла собственности для выборки оставшихся клиентов должен быть числовым.')
    else:
        print('Требование не выполнено: тип данных балла собственности для выборки ушедших клиентов должен быть числовым.')
else:
    print('Требование не выполнено: выборки ушедших и оставшихся клиентов должны содержать данные.')
```

**Вывод и бизнес-интерпретация:** p-значение составило `3.19e-152`, что многократно ниже порогового значения `α = 0.05`. Мы **отвергаем нулевую гипотезу**. Распределения балла собственности у ушедших и оставшихся клиентов статистически различаются.

**Ценность для бизнеса:** Балл собственности (`equity`) является **мощнейшим маркером оттока**. Анализ показывает, что ушедшие клиенты концентрируются в диапазоне 3-5 баллов собственности (средний класс с имуществом/кредитами), в то время как оставшиеся клиенты чаще имеют 0 баллов или высокий балл. Это означает, что банк теряет специфическую прослойку клиентов — `средний класс с обязательствами`. Следовательно, **балл собственности должен быть положен в основу сегментации** при разработке стратегий удержания клиентов.

```python
# функция для фильтрации данных
def segment(df, products=None, equity=None, score=None, balance_range=None, 
            estimated_salary_range=None, age=None, gender=None, last_activity=None):
    filters = []
    
    if products is not None:
        filters.append(f'(products >= {products[0]}) & (products <= {products[1]})')
    if equity is not None:
        filters.append(f'(equity >= {equity[0]}) & (equity <= {equity[1]})')
    if score is not None:
        filters.append(f'(score >= {score[0]}) & (score <= {score[1]})')
    if balance_range is not None:
        filters.append(f'(balance >= {balance_range[0]}) & (balance <= {balance_range[1]})')
    if estimated_salary_range is not None:
        filters.append(f'(estimated_salary >= {estimated_salary_range[0]}) & (estimated_salary <= {estimated_salary_range[1]})')
    if age is not None:
        filters.append(f'(age >= {age[0]}) & (age <= {age[1]})')
    if gender is not None:
        filters.append(f'(gender >= {gender[0]}) & (gender <= {gender[1]})')
    if last_activity is not None:
        filters.append(f'(last_activity >= {last_activity[0]}) & (last_activity <= {last_activity[1]})')

    filtered_df = df.query(' & '.join(filters))
    return filtered_df
```

## Шаг 6. Выводы

Региональное отделение "Метанпромбанк" проводит исследование по сегментации клиентов по количеству потребления банковских услуг. В нашу задачу также входит потретирование клиентов, которые склонны уходить из банка. Ситуация, в которой клиент банка прекращает пользоваться услугами и переходит к конкурентам, или вовсе прекращает взаимодействие с учреждением, в банковской сфере обозначает процесс ухода или потери клиентов банка и называется `оттоком`.

Сам по себе отток клиентов является очень серьёзной проблемой для банков, поскольку привлечение новых клиентов требует значительных затрат на маркетинг и рекламу, а также на создание новых продуктов и услуг. Кроме того, удержание существующих клиентов является важным фактором для повышения доходности банка и поддержания его стабильности.

Причины оттока клиентов из банка могут быть разнообразными. Некоторые из них включают неудовлетворительное качество обслуживания, высокие комиссии и платежи, неудобство использования банковских услуг, низкую процентную ставку по депозитам или кредитам, наличие лучших предложений у конкурентов, изменение финансовых потребностей или жизненных обстоятельств клиента.

В "Метанпромбанк" зафиксировали тревожно высокий общий уровень оттока клиентов и инициировали исследование, поскольку они активно и систематически работают над снижением оттока клиентов. Как мы предполагаем, ранее сотрудниками "Метанпромбанка" были разработаны некоторые программы лояльности, уделялось внимание качеству обслуживания клиентов, проводился периодический анализ данных для прогнозирования оттока и принятия соответствующих мер. Однако, на текущий момент тревожная 18%-ая планка начала плавно подниматься. Поэтому нам в целях проведения стороннего исследования был предоставлен оригинальный массив данных по клиентам "Метанпромбанк". Наши действия по исследованию указанных данных изложены далее по шагам:

**`ШАГ 1.`** Нами был **импортирован ряд библиотек**, в том числе `pandas` для работы с массивами данных, `matplotlib` и `seaborn` для визуализации полученных результатов. Осуществлена загрузка исходных данных методом `read_csv`. В работу поступил датафрейм `bank_scrooge.csv`, содержащий информацию о клиентах «Метанпромбанка». На основании результатов последовательного применения методов `min()`, `max()`, `unique()`, `info()` и `describe()` мы получили первичную информацию для статистического анализа. Так, исходный датафрейм содержит 10000 строк в разрезе 12 признаков. Клиентов, которые пользуются 1-2 базовыми и самыми популярными продуктами: счётом в банке и кредитной картой - 84.5%. Это в совокупности постоянные клиенты банка, которые при любых условиях и обстоятельствах остаются верны "Метанпромбанку". 

Мы также изучили относительные значения признаков баланса счёта, оценочной заработной платы, кредитного скора, балла собственности клиента в разрезе признака оттока и количества банковских продуктов. Дополнительные исследования по кредитным картам, активности клиента, возрасту, гендерному признаку позволили нам выявить основной массив характерных признаков, которые  в исчерпывающей мере описывают портреты клиентов и паттерны их поведения.

**`ШАГ 4.`** Выполнена **проверка статистических гипотез** о различии **распределений** возраста между клиентами, пользующимися одним и двумя продуктами банка, а также о различии **распределений** балла собственности между ушедшими и оставшимися клиентами. 

Для обеих гипотез были подготовлены соответствующие выборки данных и проведены проверки на нормальность распределения и равенство дисперсий. Поскольку данные не соответствовали требованиям параметрических тестов, было принято обоснованное решение об использовании непараметрического `критерия Манна-Уитни`, который оценивает различия между распределениями (рангами) двух независимых выборок.

По результатам первой гипотезы не удалось выявить статистически значимого различия между **распределениями возраста** клиентов с 1 и 2 продуктами. Это указывает на то, что возрастные структуры этих групп статистически неотличимы. Следовательно, возраст был отклонён как характерный признак для разделения этих групп при сегментировании.

По результатам второй гипотезы было выявлено статистически значимое различие в **распределениях балла собственности** между ушедшими и оставшимися клиентами. Это послужило строгим математическим подтверждением значимости признака `equity` (балл собственности) как ключевого маркера оттока и основы для дальнейшей сегментации клиентов банка.

**`ШАГ 5.`** На основе данных о количестве потребляемых продуктов выполнена **сегментация клиентов.** Мы сверились с общим ходом исследования и выделили наиболее вероятные банковские продукты:
1. Открытие банковского счёта. Большинство банков предлагает различные виды счетов, такие как текущие счета, сберегательные счета или депозитные счета, которые позволяют клиентам хранить и управлять своими финансами.
2. Выпуск банковских карт. Кредитные и дебетовые карты являются одним из самых популярных банковских предложений. Карты позволяют клиентам осуществлять покупки и снятие наличных, а также предлагают дополнительные привилегии, такие как программа лояльности или кэшбэк. Кредитная карта банка имеется у 67,7% клиентов "Метанпромбанка".
3. Кредиты и ипотеки. Банки предлагают клиентам-физическим лицам возможность получения кредитов на различные цели, включая автомобильные кредиты, потребительские кредиты и ипотечные кредиты на покупку недвижимости. Стоит отметить, что 60,5% всех клиентов банка пользуются 1-2 банковскими продуктами.
4. Страхование. Банки могут предлагать клиентам страховые продукты, такие как автомобильное страхование, страхование недвижимости, страхование жизни и другие виды страхования. Это позволяет клиентам защитить свое имущество и финансовое будущее.
5. Инвестиционные продукты. Многие банки предлагают клиентам возможность инвестирования своих средств в различные финансовые инструменты, такие как акции, облигации, паи инвестиционных фондов и другие. Это позволяет клиентам увеличивать свой капитал и получать доход от инвестиций. Этим видом продуктов активно пользуется высший сегмент среднего социального класса Ярославской области (бизнесмены и предприниматели в возрасте от 32 до 45 лет).

В соответствии с поставленной задачей по формированию 3-5 укрупнённых сегментов клиентов численностью от 300 до 2000 человек, нами был сформирован фильтр-макет требований к данным. В отношении каждого сегмента были сформированы комплексные адресные предложения:

<b>Сегмент №1. "Обыватели":</b>
* Предложить клиентам с низким и нестабильным уровнем дохода простые и доступные финансовые продукты и услуги, которые соответствуют их потребностям. Например, это могут быть расчётные счета без минимального остатка или низкими комиссиями, микрокредиты или недорогие потребительские кредиты.

Это может быть в виде бесплатных онлайн-курсов, семинаров или консультаций с финансовыми специалистами. Приложив некоторые усилияпо улучшению уровня образования в финансах своих "низовых" клиентов сегодня, через год-два банк имеет шанс получить клиента с другим уровнем дохода (в сторону роста).

* Разработка или адаптация действующей версии мобильного приложения до приложения с простым и интуитивно понятным интерфейсом, разместить банкоматы в доступных местах с повышенным трафиком и рядом с социальными учреждениями, киоски для пополнения счетов или платёжных терминалов. В том числе, внедрить малораспространённую, но востребованную опцию моментальной отмены платежа.

* И, наконец, самое простое и самое действенное: маркетинговая коммуникация с клиентами этого сегмента должна быть организована на понятном обывателю языке. Из таких мелочей складывается итоговый портрет учреждения, который потом посетители так или иначе транслируют своему окружению. А это, в свою очередь, уже реклама. Какой будет её качество и содержание - зависит от работников банка на месте.

<b>Сегмент №2. "Постоянные клиенты":</b>
* Усилить дифференциацию услуг и внимательнее посмотреть на потребности представителей этого сегмента. Это могут быть кредитные продукты с низкими процентными ставками, программы по шкалам кредитного скоринга или программы лояльности со специальными бонусами и скидками.

Такие программы, ориентированные на самую широкую публику, помогут клиентам лучше понимать специальные термины, финансовые процессы и, в конечном итоге, принимать взвешенные и обоснованные финансовые решения. Так или иначе, это поможет сократить число клиентов, которые вследствие недостаточной информированности оформили кредитную карту вместо, например, только расчётного счёта. Или открыли счёт в рублях, а на самом деле им нужен был мультивалютный депозит и для рублей, и тенге, и юаней. Но досадная ошибка уже сделана, рублёвый счёт открыт и простаивает в бездействии. А следом за этим неизбежно наступает отток клиента.

Неотложная помощь с самыми запутанными финансовыми вопросами высоко ценится клиентами, и, по завершении взаимодействия с работником банка, создаёт доверие и запас уверенности в услугах банка, а это - уже готовая почва для углубления взаимоотношений и возможность сократить отток клиентов.

* Улучшение и адаптация программ управление финансовыми рисками - предоставить клиентам инструменты и ресурсы для управления и снижения своих финансовых рисков. Это может быть образовательная информация о планировании бюджета, сберегательных счетах или программы страхования.
* Предложить программы лояльности с дополнительными бонусами и вознаграждениями. Это может быть накопление баллов за использование банковских карт или предоставление специальных условий при платежах по кредитам или ипотеке.

К примеру, можно создать персонализированные письма или SMS-сообщения с предложениями о новых продуктах или услугах, которые могут быть полезны клиенту. И это - как раз одна из задач нашего исследования.

Стоит немного перестроиться и сформировать сбалансированную программу, которая позволит клиентам получать дополнительные бонусы, скидки или привилегии при использовании продуктов или услуг банка. Потенциально, на рынке имеются компании, с которыми можно скооперироваться и создать совместный продукт или услугу, полезную, нужную и востребованную клиентом. Например, можно сотрудничать с автодилерами или туристическими агентствами для предоставления специальных условий на автокредиты или туры.

Это может включать использование высоких стандартов безопасности, многоуровневую аутентификацию, шифрование данных, а также консультации по вопросам приватности и защиты информации.

* Пригласить клиентов из этого сегмента к участию в благотворительных и общественных проектах "Метанпромбанка". Это направление в последнее десятилетие только набирает обороты и сама форма часто-государственного партнёрства становится всё более востребованной. Банк может поддерживать и финансировать благотворительные и социальные проекты, которые интересны клиентам из этого сегмента. Это может быть спонсорство культурных или спортивных событий, поддержка образовательных программ или вовлечение в программы социальной ответственности.
* Заканчиваем опять на простом человеческом уровне: любите своих клиентов! Удобство и высокий уровень сервиса дадут возможность использовать широкий спектр удобных каналов обслуживания. Это могут быть удобные цифровые каналы, мобильные приложения или онлайн-банкинг с расширенным набором функций и возможностей.

Подводя итоги вышесказанному, хотим заметить, что удержание клиентов в банковском деле — это основа основ успешного развития. Постоянные клиенты обеспечивают поступление регулярных доходов через использование банковских услуг, таких как счета, кредиты, инвестиции и другие финансовые продукты. Рост активов, в свою очередь, позволяет банку расширять свои возможности предоставления услуг и конкурировать на рынке более эффективно.

Клиенты, удовлетворенные качеством обслуживания и предоставляемыми услугами, склонны оставаться верными своему банку и рекомендовать его другим людям. Положительная репутация привлекает новых клиентов и способствует развитию банка. Кроме того, лояльность клиентов также снижает риски оттока и потери клиентской базы конкурентам.

Исследование оттока клиентов позволит банку глубже понять потребности своей аудитории. Взаимодействие с постоянными клиентами дает банку возможность лучше понимать их финансовые цели, предпочтения и потребности. Это помогает банку создавать персонализированные продукты и услуги, адаптированные под индивидуальные потребности клиентов, что в конечном итоге способствует удовлетворенности и удержанию клиентов.

На основании общераспространённой практики, удержание клиентов экономически выгоднее, чем привлечение новых. Привлечение новых клиентов требует значительных маркетинговых и рекламных затрат, в то время как сохранение существующих клиентов может достигаться через качественное обслуживание, персонализацию и предоставление дополнительных преимуществ.

Таким образом, удержание клиентов является фундаментальным стратегически важным аспектом успешного развития банковского дела в долгосрочной перспективе. Оно способствует стабильности и увеличению доходов, формированию крепких и гибких отношений с целевой аудиторией, повышению лояльности, сокращению потерь и развитию лояльности и репутации банка.

На этом полагаем цель исследования достигнутой, а проектную работу — выполненной.
